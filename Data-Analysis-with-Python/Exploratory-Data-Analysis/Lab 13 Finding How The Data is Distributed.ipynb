### Step 1: Import Libraries and Load Data


- Import the `pandas`, `matplotlib.pyplot`, and `seaborn` libraries.


- You will begin with loading the dataset. You can use the pyfetch method if working on JupyterLite. Otherwise, you can use pandas' read_csv() function directly on their local machines or cloud environments.


```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Display the first few rows of the dataset
df.head()
```

### Step 2: Examine the Structure of the Data


- Display the column names, data types, and summary information to understand the data structure.

- Objective: Gain insights into the dataset's shape and available variables.


```python
print("Dataset shape:", df.shape)

display(df.dtypes.to_frame(name="Data type"))
df.info()
display(df.describe(include="all").T)
```

### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
# Show missing-value counts for columns that contain missing data
missing_counts = df.isna().sum().sort_values(ascending=False)
display(missing_counts[missing_counts > 0].to_frame("Missing values"))

# Keep rows that have values for the columns used in the next steps
key_columns = ["Employment", "JobSat", "YearsCodePro"]
rows_before = len(df)

df = df.dropna(subset=key_columns).copy()

print("Rows removed:", rows_before - len(df))
print("Rows remaining:", len(df))
```

### Step 4: Analyze Key Columns


- Examine key columns such as `Employment`, `JobSat` (Job Satisfaction), and `YearsCodePro` (Professional Coding Experience).

- **Instruction**: Calculate the value counts for each column to understand the distribution of responses.



```python
for column in ["Employment", "JobSat", "YearsCodePro"]:
    print(f"\nValue counts for {column}:")
    display(df[column].value_counts().to_frame("Count"))
```

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
import matplotlib.pyplot as plt
import pandas as pd

job_sat_counts = df["JobSat"].value_counts()

# Put numeric satisfaction scores in ascending order
if pd.api.types.is_numeric_dtype(df["JobSat"]):
    job_sat_counts = job_sat_counts.sort_index()

plt.figure(figsize=(8, 8))
plt.pie(
    job_sat_counts.values,
    labels=job_sat_counts.index.astype(str),
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Distribution of Job Satisfaction")
plt.axis("equal")
plt.show()

most_common = job_sat_counts.idxmax()
share = job_sat_counts.max() / job_sat_counts.sum() * 100

print(f"Most common JobSat response: {most_common} ({share:.1f}% of responses)")
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
def get_language_counts(column):
    return (
        df[column]
        .dropna()
        .astype(str)
        .str.split(";")
        .explode()
        .str.strip()
        .replace("", pd.NA)
        .dropna()
        .value_counts()
    )

worked = get_language_counts("LanguageHaveWorkedWith").rename("Worked with")
wanted = get_language_counts("LanguageWantToWorkWith").rename("Want to work with")

language_comparison = pd.concat([worked, wanted], axis=1).fillna(0)

top_languages = (
    language_comparison
    .nlargest(10, "Worked with")
    .sort_values("Worked with")
)

display(top_languages)

top_languages.plot(kind="barh", figsize=(10, 6))
plt.title("Programming Languages: Used vs. Wanted")
plt.xlabel("Number of respondents")
plt.ylabel("Programming language")
plt.tight_layout()
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
top_countries = df["Country"].value_counts().head(10).index

remote_by_country = pd.crosstab(
    df["Country"],
    df["RemoteWork"],
    normalize="index"
).loc[top_countries] * 100

display(remote_by_country.round(1))

fig, ax = plt.subplots(figsize=(12, 6))
sns.heatmap(
    remote_by_country,
    annot=True,
    fmt=".1f",
    cmap="Blues",
    cbar_kws={"label": "Percentage"},
    ax=ax
)
ax.set_title("Remote Work Distribution by Country (%)")
ax.set_xlabel("Remote work arrangement")
ax.set_ylabel("Country")
fig.subplots_adjust(left=0.2, bottom=0.2)
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
years = df["YearsCodePro"].astype("string").str.strip()
years = years.replace({
    "Less than 1 year": "0.5",
    "More than 50 years": "51"
})

correlation_data = pd.DataFrame({
    "YearsCodePro": pd.to_numeric(years, errors="coerce"),
    "JobSat": pd.to_numeric(df["JobSat"], errors="coerce")
}).dropna()

pearson_corr = correlation_data["YearsCodePro"].corr(
    correlation_data["JobSat"],
    method="pearson"
)

print(f"Pearson correlation: {pearson_corr:.3f}")
```

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
employment_education = pd.crosstab(
    df["EdLevel"],
    df["Employment"],
    normalize="index"
) * 100

display(employment_education.round(1))

fig, ax = plt.subplots(
    figsize=(13, max(6, len(employment_education) * 0.6))
)

employment_education.plot(kind="barh", stacked=True, ax=ax)
ax.set_xlim(0, 100)
ax.set_title("Employment Type by Education Level (%)")
ax.set_xlabel("Percentage")
ax.set_ylabel("Education level")
ax.legend(title="Employment", bbox_to_anchor=(1.02, 1), loc="upper left")

fig.subplots_adjust(left=0.35, right=0.78)
plt.show()
```
