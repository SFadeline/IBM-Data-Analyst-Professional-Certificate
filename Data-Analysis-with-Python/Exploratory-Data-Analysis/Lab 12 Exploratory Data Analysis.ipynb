#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

critical_cols = ["Employment", "JobSat", "RemoteWork"]

# Check missing values before handling them
print("Missing values before:")
print(df[critical_cols].isna().sum())

# Fill categorical columns with their most frequent value
for col in ["Employment", "RemoteWork"]:
    modes = df[col].mode(dropna=True)
    if not modes.empty:
        df[col] = df[col].fillna(modes.iloc[0])

# JobSat is used for the satisfaction analysis, so remove rows without a score
df["JobSat"] = pd.to_numeric(df["JobSat"], errors="coerce")
df = df.dropna(subset=["JobSat"]).copy()

print("\nMissing values after:")
print(df[critical_cols].isna().sum())
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
years = df["YearsCodePro"].astype("string").str.strip()

years = years.replace({
    "Less than 1 year": "0.5",
    "More than 50 years": "51"
})

df["YearsCodePro_numeric"] = pd.to_numeric(years, errors="coerce")

df["ExperienceRange"] = pd.cut(
    df["YearsCodePro_numeric"],
    bins=[-np.inf, 5, 10, 20, np.inf],
    labels=["0–5 years", "5–10 years", "10–20 years", "Over 20 years"]
)

median_by_experience = (
    df.groupby("ExperienceRange", observed=False)["JobSat"]
      .median()
      .dropna()
)

display(median_by_experience.rename("Median Job Satisfaction").to_frame())

plt.figure(figsize=(8, 5))
plt.bar(median_by_experience.index.astype(str), median_by_experience.values)
plt.xlabel("Professional coding experience")
plt.ylabel("Median job satisfaction")
plt.title("Median Job Satisfaction by Experience Range")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="JobSat", order=sorted(df["JobSat"].dropna().unique()))
plt.xlabel("Job satisfaction score")
plt.ylabel("Number of respondents")
plt.title("Distribution of Job Satisfaction")
plt.tight_layout()
plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
remote_df = df[["RemoteWork", "Employment"]].dropna().copy()

# Overall remote work distribution
plt.figure(figsize=(8, 5))
sns.countplot(
    data=remote_df,
    x="RemoteWork",
    order=remote_df["RemoteWork"].value_counts().index
)
plt.title("Remote Work Preferences")
plt.xlabel("Remote work arrangement")
plt.ylabel("Number of respondents")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

# Remote work preference by employment type
remote_by_employment = pd.crosstab(
    remote_df["Employment"],
    remote_df["RemoteWork"]
)
display(remote_by_employment)
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
country_name = "Malaysia"

country_data = df.loc[
    df["Country"].eq(country_name),
    "LanguageHaveWorkedWith"
].dropna()

top_languages = (
    country_data.str.split(";")
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

display(top_languages.rename("Respondents").to_frame())

plt.figure(figsize=(9, 5))
top_languages.sort_values().plot(kind="barh")
plt.title(f"Top Programming Languages in {country_name}")
plt.xlabel("Number of respondents")
plt.ylabel("Programming language")
plt.tight_layout()
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
years = df["YearsCodePro"].astype("string").str.strip()
years = years.replace({
    "Less than 1 year": "0.5",
    "More than 50 years": "51"
})

analysis_df = pd.DataFrame({
    "YearsCodePro": pd.to_numeric(years, errors="coerce"),
    "JobSatPoints_1": pd.to_numeric(df["JobSatPoints_1"], errors="coerce")
}).dropna()

print("Correlation:", analysis_df["YearsCodePro"].corr(
    analysis_df["JobSatPoints_1"]
))

plt.figure(figsize=(8, 5))
plt.scatter(
    analysis_df["YearsCodePro"],
    analysis_df["JobSatPoints_1"],
    alpha=0.25,
    s=12
)
plt.title("Professional Coding Experience and Job Satisfaction")
plt.xlabel("Years of professional coding experience")
plt.ylabel("Job satisfaction points")
plt.tight_layout()
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
import textwrap
import matplotlib.pyplot as plt
import pandas as pd

# Employment percentages within each education level
education_employment = pd.crosstab(
    df["EdLevel"],
    df["Employment"],
    normalize="index"
) * 100

display(education_employment.round(1))

# Keep the six most common employment types; combine the rest
top_employment = df["Employment"].value_counts().head(6).index
plot_data = education_employment[list(top_employment)].copy()

other_cols = [
    col for col in education_employment.columns
    if col not in top_employment
]
if other_cols:
    plot_data["Other"] = education_employment[other_cols].sum(axis=1)

# Wrap long education labels so they take up less space
plot_data.index = [
    textwrap.fill(str(label), width=32)
    for label in plot_data.index
]

fig, ax = plt.subplots(
    figsize=(13, max(6, len(plot_data) * 0.6))
)

plot_data.plot(kind="barh", stacked=True, ax=ax)

ax.set_xlim(0, 100)
ax.set_title("Employment Type by Education Level (%)")
ax.set_xlabel("Percentage")
ax.set_ylabel("Education level")
ax.legend(title="Employment type", bbox_to_anchor=(1.02, 1), loc="upper left")

# Reserve room for the long labels and the legend
fig.subplots_adjust(left=0.35, right=0.78, top=0.92, bottom=0.10)
plt.show()
```
